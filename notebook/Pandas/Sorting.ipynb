```python
import numpy as np
import pandas as pd
sr_scores = pd.Series([90, 85, 78, 92, 85], index=['Alice', 'Bob', 'Charlie', 'David', 'Eva'])
df_sales = pd.DataFrame({
    "order_id": [1, 2, 3, 4, 5, 6, 7, 8],
    "product": ["Laptop", "Phone", "Tablet", "Laptop", "Phone", "Tablet", "Laptop", "Phone"],
    "price": [1000, 500, 300, 1200, 550, 320, 1100, 600],
    "quantity": [1, 2, 3, 1, 2, 1, 1, 2],
    "order_date": [
        "2024-01-01", "2024-01-02", "2024-01-03", "2024-01-04",
        "2024-01-05", "2024-01-06", "2024-01-07", "2024-01-08"
    ]
})
df_employees = pd.DataFrame({
    "employee_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank", "Grace", "Helen"],
    "department": ["HR", "IT", "IT", "Finance", "HR", "IT", "Finance", "HR"],
    "age": [25, 32, 29, 41, 35, 28, 45, 30],
    "salary": [50000, 70000, 65000, 90000, 62000, 72000, 88000, 58000],
    "city": ["Hanoi", "HCMC", "Hanoi", "Danang", "HCMC", "Hanoi", "Danang", "HCMC"],
    "is_active": [True, True, False, True, True, False, True, True]
})
```

```python
dept_order = pd.Categorical(
    df_employees["department"],
    categories= ['IT','HR',"FInance"],
    ordered= True
)

df1 = df_employees.assign(
    department_order = dept_order
).sort_values(by = "department_order"
).drop(columns = "department_order")
df1
```

```python
df_sales.assign(
    product_order = pd.Categorical(
        df_sales['product'],
        categories= ['Laptop','Tablets','Phone'],
        ordered= True
    )
).sort_values(
    by = 'product_order'
).drop(
    columns = 'product_order'
)
```

```python
df_sales.assign(
    result = df_sales['price']*df_sales['quantity']
).sort_values(
    by = ['result', 'quantity'], ascending= [False,False]
).drop(
    columns = 'result'
)
```

```python
df_employees.assign(
    department_order = pd.Categorical(
        df_employees['department'],
        categories= ['Finance', 'IT', 'HR'],
        ordered = True
    )
).sort_values(
    by = ['is_active', 'department_order','salary','age'], ascending= [False, True, False, True]
).drop(
    columns = 'department_order'
)
```
