#Create a series

```python
import pandas as pd
a = pd.Series(['a',1,2,3],index =['a','b','c','d'])
print(a)
print(a['d'])
print(a.values)
print(a.index)
type(a)
```

#Selection

```python
print(a['a'])
a['a'] = 6
print(a[['a','b','c']])
```

#Vectorized operation

```python
b = a* 2
print(b)
print(a[a > 2])
a[a > 2] = 6
print(a)
```
