## II/ Mo phong thong ke

```python
flys = Table().read_table('../../Data/united_summer2015.csv')
np.median(flys.column('Delay'))
```

```python
def one_simulation():
    return np.median(flys.sample(1000).column('Delay'))

repitition = 5000
result = make_array()
for i in np.arange(repitition):
    result = np.append(result,one_simulation())
simu = Table().with_column(
    'Median',result
)
simu.hist(bins = np.arange(0.5,4.6,1))
```
