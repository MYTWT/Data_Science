```python
import numpy as np
from datascience import *
import matplotlib.pyplot as plots
%matplotlib inline
```

kích thước mẫu là 100, bằng kích thước của nhóm chuyên gia. Phân phối mà chúng ta sẽ lấy mẫu là phân phối trong quần thể các bồi thẩm viên đủ điều kiện. Vì 26% trong số họ là người da đen, chúng ta sẽ lấy mẫu từ phân phối được xác định bởi tỷ lệ [0,26, 0,74].

```python
sample_size = 100
sample_proportions(sample_size, [0.26,0.74]).item(0)
```

```python
def one_simulated_member():
    return sample_size*sample_proportions(sample_size,[0.26,0.74]).item(0)
one_simulated_member()
```

```python
count = make_array()
for i in np.arange(10000):
    count = np.append(count,one_simulated_member())
Member = Table().with_column(
    'Count',count
)
Member.hist(bins = np.arange(5.5,44.6,1))
```

Mo phong TVD cua hai phan phoi

```python
ditribution1 = [0.15,0.18,0.54,0.12,0.01]
def one_simulated_TVD(distribu1,distribu2):
    return np.sum(np.abs(distribu1 - distribu2))/2
```

```python
repitation = 10000
result = make_array()
for i in np.arange(repitation):
    result = np.append(result,one_simulated_TVD(ditribution1,sample_proportions(1453,ditribution1)))
result
res = Table().with_column(
    'TVD',result
)
res.hist(bins = np.arange(0,0.2,0.005))
plots.xlabel('TVD')
plots.ylabel('Frequency')
```
